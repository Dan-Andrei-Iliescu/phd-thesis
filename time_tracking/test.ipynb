```python
import pandas as pd
```

```python
df = pd.read_csv("report.csv")
grouped = df[df['Project']=='Papercup'] \
    .groupby(['Entry'])[['Logged', 'Scheduled']].sum() \
    .sort_values('Logged', ascending=False)
print(grouped)
print("Logged total: ", grouped['Logged'].sum(), ", Scheduled total: ", grouped['Scheduled'].sum())
```

```python
import pandas as pd
from datetime import datetime, time, timedelta
import time
import ipywidgets as widgets
```

```python
TASK_FILE = "tasks.csv"
RECORD_FILE = "record.csv"
DT_FORMAT = "%Y-%m-%d %H:%M"
T_FORMAT = "%H:%M"

def cast_timedelta(x):
    if not x:
        return x
    
    h, m = map(int, x.split(':'))
    return timedelta(hours=h, minutes=m)


def cast_datetime(x):
    if not x:
        return x
    
    return datetime.strptime(x, DT_FORMAT)

COL_DTYPES = {
    'task_id':str,
    'parent_id':str,
    'start_date': cast_datetime,
    'end_date': cast_datetime,
    'deadline': cast_datetime,
    'is_completed':bool,
    'est_time': cast_timedelta,
    'log_time': cast_timedelta,
    'description':str
}
```

```python
def get_tasks():
    return pd.read_csv(TASK_FILE, converters=COL_DTYPES)

def set_task(task_df, new_task):
    df = pd.DataFrame([new_task], columns=task_df.columns)
    task_df = pd.concat([task_df, df])
    task_df.to_csv(TASK_FILE, index=False)

def create_task():
    task_df = get_tasks()
    new_task = []
    for col_name in task_df.columns:
        if col_name == "task_id":
            val = task_df[col_name].max() + 1
        elif col_name == "is_completed":
            val = False
        elif col_name == "start_date":
            val = date.today()
        elif col_name == "end_date":
            val = date.today()
        elif col_name == "deadline":
            print("yes")
            val = widgets.DatePicker(
                description='Pick a Date',
                disabled=False
            )
        else:
            val = input("Set " + col_name)
        new_task.append(val)
    set_task(task_df, new_task)
```
